# Q. The Maximum Entropy Distribution under Mean and Variance Constraints

Consider the random variable $X \in (-\infty, \infty)$. Let:

1. **The Mean:** $\mathbb{E}[X] = \mu$
2. **The Variance:** $\mathbb{E}[(X-\mu)^2] = \sigma^2$

Show that the law of $X$ that maximizes entropy is given by:

$$X \sim \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left( -\frac{(x - \mu)^2}{2\sigma^2} \right)$$

Answer:

To find the probability density function $p(x)$ that maximizes the differential entropy:$$\max_{p} h(X) = -\int_{-\infty}^{\infty} p(x) \ln p(x) \, dx$$Subject to three operational constraints:Normalization: $\int_{-\infty}^{\infty} p(x) \, dx = 1$Mean constraint: $\int_{-\infty}^{\infty} x p(x) \, dx = \mu$Variance constraint: $\int_{-\infty}^{\infty} (x-\mu)^2 p(x) \, dx = \sigma^2$We construct the functional optimization objective using Lagrange multipliers $\lambda_0, \lambda_1, \lambda_2$:$$\mathcal{L}(p, \lambda_0, \lambda_1, \lambda_2) = -\int_{-\infty}^{\infty} p(x) \ln p(x) \, dx - \lambda_0 \left(\int_{-\infty}^{\infty} p(x) \, dx - 1\right) - \lambda_1 \left(\int_{-\infty}^{\infty} x p(x) \, dx - \mu\right) - \lambda_2 \left(\int_{-\infty}^{\infty} (x-\mu)^2 p(x) \, dx - \sigma^2\right)$$Taking the functional derivative with respect to $p(x)$ and setting it to zero:$$\frac{\delta \mathcal{L}}{\delta p(x)} = -1 - \ln p(x) - \lambda_0 - \lambda_1 x - \lambda_2 (x-\mu)^2 = 0$$$$\ln p(x) = -(1 + \lambda_0) - \lambda_1 x - \lambda_2 (x-\mu)^2$$$$p(x) = \exp(-(1+\lambda_0)) \cdot \exp\left(-\lambda_1 x - \lambda_2 (x-\mu)^2\right)$$Since the variance constraint is centered directly around $\mu$, the distribution must remain structurally symmetric around $x = \mu$ to satisfy $\mathbb{E}[X] = \mu$. This requires the asymmetric linear parameter $\lambda_1 = 0$. Let $C = \exp(-(1+\lambda_0))$ denote our normalization constant:$$p(x) = C \exp\left(-\lambda_2 (x-\mu)^2\right)$$We solve for $C$ and $\lambda_2$ using our integral constraints:Normalization:$$\int_{-\infty}^{\infty} C e^{-\lambda_2(x-\mu)^2} \, dx = C \sqrt{\frac{\pi}{\lambda_2}} = 1 \implies C = \sqrt{\frac{\lambda_2}{\pi}}$$Variance Evaluation:$$\int_{-\infty}^{\infty} (x-\mu)^2 \left(\sqrt{\frac{\lambda_2}{\pi}} e^{-\lambda_2(x-\mu)^2}\right) \, dx = \sigma^2$$Using the standard identity $\int_{-\infty}^{\infty} y^2 e^{-\lambda_2 y^2} \, dy = \frac{1}{2\lambda_2}\sqrt{\frac{\pi}{\lambda_2}}$, we see:$$\sqrt{\frac{\lambda_2}{\pi}} \cdot \left[\frac{1}{2\lambda_2}\sqrt{\frac{\pi}{\lambda_2}}\right] = \frac{1}{2\lambda_2} = \sigma^2 \implies \lambda_2 = \frac{1}{2\sigma^2}$$Substituting $\lambda_2$ back into $C$ gives $C = \frac{1}{\sqrt{2\pi\sigma^2}}$. Thus, our distribution is uniquely determined as the classical Gaussian Distribution:$$X \sim \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left( -\frac{(x - \mu)^2}{2\sigma^2} \right) \quad \blacksquare$$

# Q. Asset Pricing and Portfolio Optimization.



In standard finance, models often assume asset returns follow a perfect Normal (Gaussian) distribution. However, real-world markets experience extreme events, leading to "fat tails" and skewed returns. MaxEnt allows an analyst to infer the **least-biased probability distribution** of an asset's future price based *only* on what the market currently tells us, without forcing it into a rigid Gaussian shape.

**Background:**

You are a quantitative portfolio manager evaluating a highly volatile tech stock, **Asset X**, currently trading at $\$100$. You want to determine the risk-neutral probability distribution of its price, $S_T$, in exactly one month. This distribution will allow you to correctly price exotic derivatives and optimize your portfolio's risk exposure.

**Available Market Data (The Constraints):**

You do not know the true distribution, but you can observe the following macroeconomic aggregates and market-clearing prices from the options exchange:

1. **The Risk-Free Growth:** Based on the risk-free interest rate, the expected future price of the asset must match the forward price:

$$\mathbb{E}[S_T] = \$102$$


2. **The Market Volatility:** Based on standard historical variance, the expected squared price is constrained by:

$$\mathbb{E}[S_T^2] = 10,600$$


3. **Asymmetric Market Sentiment (The "Skew"):** Due to heavy trading on out-of-the-money put options, the market is pricing in a strong downside risk. The expected value of a specific logarithmic penalty (which captures downward crash risk) is constrained to:

$$\mathbb{E}[\ln(S_T)] = 4.60$$



**Objective:**

Formulate and solve the Maximum Entropy problem to find the continuous probability density function $p(S_T)$ for $S_T > 0$.



Answers:

Formulation of the MaxEnt ProblemWe want to find the continuous risk-neutral probability density function $p(S_T)$ over the domain $S_T \in (0, \infty)$ that maximizes information uncertainty subject to our physical pricing constraints.$$\max_{p} h(p) = -\int_{0}^{\infty} p(S_T) \ln p(S_T) \, dS_T$$Subject to the Constraints:Normalization: $\int_{0}^{\infty} p(S_T) \, dS_T = 1$Risk-Free Asset Growth: $\int_{0}^{\infty} S_T p(S_T) \, dS_T = 102$Market Volatility Bounds: $\int_{0}^{\infty} S_T^2 p(S_T) \, dS_T = 10,600$Asymmetric Log-Skew Penalty: $\int_{0}^{\infty} \ln(S_T) p(S_T) \, dS_T = 4.60$Deriving the Functional FormUsing the method of Lagrange multipliers, we assign $\lambda_0$ to normalization, $\lambda_1$ to the asset mean, $\lambda_2$ to the squared asset mean, and $\lambda_3$ to the logarithmic skew constraint. Optimizing functionally yields:$$\frac{\delta}{\delta p(S_T)} \left[ -\int p(S_T)\ln p(S_T) dS_T - \lambda_0\left(\int p(S_T) dS_T - 1\right) - \lambda_1\left(\int S_T p(S_T) dS_T - 102\right) - \lambda_2\left(\int S_T^2 p(S_T) dS_T - 10600\right) - \lambda_3\left(\int \ln(S_T) p(S_T) dS_T - 4.60\right) \right] = 0$$$$-1 - \ln p(S_T) - \lambda_0 - \lambda_1 S_T - \lambda_2 S_T^2 - \lambda_3 \ln(S_T) = 0$$Solving explicitly for $p(S_T)$:$$p(S_T) = \exp(-(1+\lambda_0)) \cdot \exp\left(-\lambda_3 \ln(S_T)\right) \cdot \exp\left(-\lambda_1 S_T - \lambda_2 S_T^2\right)$$Using the logarithmic property $\exp(-\lambda_3 \ln(S_T)) = S_T^{-\lambda_3}$, and grouping the normalization factor into a scaling constant $C = \exp(-(1+\lambda_0))$, we get the exact analytical solution:$$p(S_T) = C \cdot S_T^{-\lambda_3} \exp\left(-\lambda_1 S_T - \lambda_2 S_T^2\right)$$Economic InterpretationThis result belongs to the Generalized Gamma / Modulated Gaussian distribution family.If $\lambda_2 = 0$, the distribution simplifies into a standard Gamma distribution, frequently used to model asset durations.If $\lambda_1 = 0$ and $\lambda_3 = 0$, it simplifies into a truncated normal distribution.By combining all terms ($\lambda_1, \lambda_2, \lambda_3$), this MaxEnt formulation allows the options market data to naturally introduce a heavy left tail (downside crash skew) and non-zero kurtosis without assuming a rigid Gaussian structure. The weights of the parameters $\lambda_i$ are solved numerically to match the empirical coordinates $(\$102, 10600, 4.60)$.
