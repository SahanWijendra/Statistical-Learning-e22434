1. Derivation of Finite-Sample Bias in the MLE Covariance Matrix

Let $\mathbf{X}_1, \mathbf{X}_2, \dots, \mathbf{X}_n$ be a sequence of $n$ independent and identically distributed ($i.i.d.$) random vectors in $\mathbb{R}^m$ with true population mean $\mathbb{E}[\mathbf{X}_i] = \boldsymbol{\mu}$ and latent population covariance matrix $\text{Var}(\mathbf{X}_i) = \boldsymbol{\Sigma}$.

The **Maximum Likelihood Estimator (MLE)** for the population covariance matrix is defined as:

$$\widehat{\boldsymbol{\Sigma}}_{\text{MLE}} = \frac{1}{n} \sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T$$

where $\widehat{\boldsymbol{\mu}}_n = \frac{1}{n}\sum_{i=1}^n \mathbf{X}_i$ represents the empirical sample mean vector.

### Algebraic Expansion

To evaluate the expectation of $\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}$, we decompose the deviation of each observation around the true population mean $\boldsymbol{\mu}$:

$$\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n = (\mathbf{X}_i - \boldsymbol{\mu}) - (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})$$

Substituting this central translation identity into the outer-product summation yields:

$$\sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T = \sum_{i=1}^n \left[ (\mathbf{X}_i - \boldsymbol{\mu}) - (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \right] \left[ (\mathbf{X}_i - \boldsymbol{\mu}) - (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \right]^T$$

Expanding the polynomial matrix multiplication term-by-term:

$$= \sum_{i=1}^n (\mathbf{X}_i - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T - \sum_{i=1}^n (\mathbf{X}_i - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T - \sum_{i=1}^n (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T + \sum_{i=1}^n (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T$$

Using the linear pooling property $\sum_{i=1}^n (\mathbf{X}_i - \boldsymbol{\mu}) = n(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})$, the two middle cross-product terms contract cleanly:

$$= \sum_{i=1}^n (\mathbf{X}_i - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T - n(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T - n(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T + n(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T$$

$$= \sum_{i=1}^n (\mathbf{X}_i - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T - n(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T$$

### Expectation Mapping

Applying the linear expectation operator $\mathbb{E}[\cdot]$ to the scaled sum maps directly onto the underlying population moments:

$$\mathbb{E}[\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}] = \frac{1}{n} \sum_{i=1}^n \mathbb{E}[(\mathbf{X}_i - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T] - \mathbb{E}[(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T]$$

By fundamental definition of multivariate spatial moments:

* **Individual variation:** $\mathbb{E}[(\mathbf{X}_i - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T] = \boldsymbol{\Sigma}$
* **Sample mean dispersion:** $\mathbb{E}[(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T] = \text{Var}(\widehat{\boldsymbol{\mu}}_n) = \frac{1}{n}\boldsymbol{\Sigma}$

Substituting these distributions back into the expectation formula:

$$\mathbb{E}[\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}] = \frac{1}{n} (n\boldsymbol{\Sigma}) - \frac{1}{n}\boldsymbol{\Sigma} = \boldsymbol{\Sigma} - \frac{1}{n}\boldsymbol{\Sigma} = \left(\frac{n-1}{n}\right)\boldsymbol{\Sigma}$$

> **Statistical Interpretation:** Because $\mathbb{E}[\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}] \neq \boldsymbol{\Sigma}$, the raw maximum likelihood matrix is systematically downward biased in finite samples. This occurs because calculating spatial scattering relative to an optimized *sample mean* ($\widehat{\boldsymbol{\mu}}_n$) rather than the *true mean* ($\boldsymbol{\mu}$) artificially shrinks the apparent sample dispersion envelope.

## 2. Unbiased Estimation via Bessel's Correction

To eliminate this mathematical bias, we scale the maximum likelihood matrix by the reciprocal of the finite population loss factor, $\frac{n}{n-1}$. This application yields the unique, unbiased sample covariance estimator $\mathbf{S}$:

$$\mathbf{S} = \frac{n}{n-1} \widehat{\boldsymbol{\Sigma}}_{\text{MLE}} = \frac{n}{n-1} \left( \frac{1}{n} \sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T \right) = \frac{1}{n-1} \sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T$$

Evaluating the expectation verifies the correction:

$$\mathbb{E}[\mathbf{S}] = \frac{n}{n-1} \mathbb{E}[\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}] = \frac{n}{n-1} \left(\frac{n-1}{n}\right)\boldsymbol{\Sigma} = \boldsymbol{\Sigma}$$

## Part B: Asymptotic Distributions & Slutsky's Theorem

### Mathematical Definition of Slutsky's Theorem

Let $\{\mathbf{X}_n\}$ be a sequence of random vectors and $\{\mathbf{Y}_n\}$ be a sequence of random matrices defined on a common probability space. If $\mathbf{X}_n$ converges in distribution to a random variable $\mathbf{X}$ (denoted as $\mathbf{X}_n \xrightarrow{d} \mathbf{X}$), and $\mathbf{Y}_n$ converges in probability to a constant deterministic matrix $\mathbf{C}$ (denoted as $\mathbf{Y}_n \xrightarrow{p} \mathbf{C}$), then:

* **Additive Stability:** $\mathbf{X}_n + \mathbf{Y}_n \xrightarrow{d} \mathbf{X} + \mathbf{C}$
* **Multiplicative Stability:** $\mathbf{Y}_n \mathbf{X}_n \xrightarrow{d} \mathbf{C}\mathbf{X}$
* **Inversion Stability:** $\mathbf{Y}_n^{-1} \mathbf{X}_n \xrightarrow{d} \mathbf{C}^{-1}\mathbf{X}$ *(provided $\mathbf{C}$ is non-singular)*

---

### Operational Distribution Justification

In practical structural health monitoring configurations, the true population covariance matrix $\boldsymbol{\Sigma}$ is a latent, unobservable parameter. To operationalize real-time anomalies or parametric boundaries, we must substitute the empirical sample covariance matrix $\mathbf{S}$ for $\boldsymbol{\Sigma}$. This substitution is formally justified as sample size $n \to \infty$ via the following asymptotic proof:

1. **Multivariate Central Limit Theorem:** Under the Lindeberg-Lévy Multivariate CLT, the standardized difference between the empirical mean vector $\widehat{\boldsymbol{\mu}}_n$ and the true mean $\boldsymbol{\mu}$ satisfies:

$$\sqrt{n}(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \xrightarrow{d} \mathscr{N}(\mathbf{0}, \boldsymbol{\Sigma})$$


2. **Standardization:** Multiplying the limiting distribution by the unique, symmetric inverse matrix square root $\boldsymbol{\Sigma}^{-1/2}$ maps the variance onto an isotropic, standard normal vector space:

$$\boldsymbol{\Sigma}^{-1/2} \left[ \sqrt{n}(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \right] \xrightarrow{d} \mathscr{N}(\mathbf{0}, \mathbf{I})$$


3. **Stochastic Consistency:** By the Weak Law of Large Numbers (WLLN), the unbiased sample covariance matrix $\mathbf{S}$ is a consistent estimator of $\boldsymbol{\Sigma}$. Because the inverse matrix square root function $(\cdot)^{-1/2}$ is continuous over the space of positive-definite matrices, Continuous Mapping Theorem (CMT) implies:

$$\mathbf{S} \xrightarrow{p} \boldsymbol{\Sigma} \implies \mathbf{S}^{-1/2} \xrightarrow{p} \boldsymbol{\Sigma}^{-1/2}$$


4. **Application of Slutsky's Theorem:** Combining the sequence converging in distribution (Step 2) with the sequence converging in probability to a constant matrix (Step 3) via Slutsky's multiplicative property yields:

$$\mathbf{S}^{-1/2} \left[ \sqrt{n}(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \right] \xrightarrow{d} \boldsymbol{\Sigma}^{-1/2} \cdot \mathscr{N}(\mathbf{0}, \boldsymbol{\Sigma}) = \mathscr{N}(\mathbf{0}, \mathbf{I})$$



### Diagnostic Conclusion

Because the observable mapping converges cleanly to a standard Gaussian space, we can transform the localized coordinate parameters back onto the empirical scale:

$$\sqrt{n}(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \xrightarrow{a} \mathscr{N}(\mathbf{0}, \mathbf{S}) \implies \widehat{\boldsymbol{\mu}}_n \xrightarrow{a} \mathscr{N}\left(\boldsymbol{\mu}, \frac{1}{n}\mathbf{S}\right)$$

> **Operational Insight:** This result provides the formal mathematical validation for substituting the sample covariance matrix $\mathbf{S}$ into your runtime diagnostic layers. It guarantees that the geometric stability of your anomaly detection thresholds and confidence envelopes remains asymptotically uncompromised despite using estimated parameters.

```python
import numpy as np
import pandas as pd
import scipy.stats as stats

# --- DO NOT MODIFY: Synthetic Strain Sensor Data Generator ---
np.random.seed(42)
n_samples = 5000
n_features = 3

# Generating base stationary multivariate normal noise
base_data = np.random.multivariate_normal(
    mean=[0.5, -0.2, 1.1],
    cov=[[0.09, 0.02, 0.01],
         [0.02, 0.06, 0.03],
         [0.01, 0.03, 0.05]],
    size=n_samples
)

# Injecting a subtle, hidden structural drift in the final 1,000 snapshots
base_data[4000:, 0] += 0.015  # Sensor 1 drift
base_data[4000:, 2] -= 0.010  # Sensor 3 drift

df_strain = pd.DataFrame(base_data, columns=['Strain_Ch1', 'Strain_Ch2', 'Strain_Ch3'])
# ----------------------------------------------------------------

def verify_first_moment_homogeneity(df: pd.DataFrame, g_chunks: int = 5) -> dict:
    """
    Partitions the dataset into g chunks and evaluates first-moment homogeneity
    via Wilks' Lambda and Bartlett's Chi-Square asymptotic transformation.
    """
    n, m = df.shape
    n_j = n // g_chunks  # Size of each chunk (1000)
    
    X = df.values
    global_mean = np.mean(X, axis=0)
    
    # Initialize Within (W) and Between (B) matrices
    W = np.zeros((m, m))
    B = np.zeros((m, m))
    
    # Process blocks
    for j in range(g_chunks):
        chunk = X[j * n_j : (j + 1) * n_j, :]
        chunk_mean = np.mean(chunk, axis=0)
        
        # Within-chunk scatter contribution
        chunk_deviations = chunk - chunk_mean
        W += np.dot(chunk_deviations.T, chunk_deviations)
        
        # Between-chunk scatter contribution
        mean_diff = (chunk_mean - global_mean).reshape(-1, 1)
        B += n_j * np.dot(mean_diff, mean_diff.T)
        
    # Calculate Determinants and Wilks' Lambda
    det_W = np.linalg.det(W)
    det_Total = np.linalg.det(B + W)
    wilks_lambda = det_W / det_Total
    
    # Bartlett's Chi-Square asymptotic transformation
    # Formula: -(n - 1 - (m + g)/2) * ln(Lambda)
    multiplier = -(n - 1 - (m + g_chunks) / 2.0)
    chi2_calc = multiplier * np.log(wilks_lambda)
    
    # Degrees of freedom: m * (g - 1)
    df_chi2 = m * (g_chunks - 1)
    p_value = 1.0 - stats.chi2.cdf(chi2_calc, df_chi2)
    
    return {
        "Wilks_Lambda": float(wilks_lambda),
        "Bartlett_Chi2": float(chi2_calc),
        "Degrees_of_Freedom": int(df_chi2),
        "p_value": float(p_value)
    }

# Execute completed framework
results = verify_first_moment_homogeneity(df_strain, g_chunks=5)
print(results)
```

## 1. Subspace Orthogonalization & Covariance Diagonalization

### Step-by-Step Proof of Covariance Diagonalization

We are given the centered vector $\widetilde{\mathbf{X}}_i = (\mathbf{X}_i - \boldsymbol{\mu})$ with $\mathbb{E}[\widetilde{\mathbf{X}}_i \widetilde{\mathbf{X}}_i^T] = \boldsymbol{\Sigma}$. The spectral decomposition of the symmetric population covariance matrix is $\boldsymbol{\Sigma} = \mathbf{P} \mathbf{\Lambda} \mathbf{P}^T$, where $\mathbf{P}$ is an orthogonal matrix ($\mathbf{P}^T\mathbf{P} = \mathbf{P}\mathbf{P}^T = \mathbf{I}$) of eigenvectors, and $\mathbf{\Lambda}$ is the diagonal matrix of eigenvalues.

The orthogonal projection mapping is defined as:


$$\mathbf{Z}_i = \mathbf{P}^T \widetilde{\mathbf{X}}_i$$

To find the theoretical covariance matrix of the transformed score vector $\mathbf{Z}_i$, we evaluate its outer product expectation:


$$\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \mathbb{E}\left[ (\mathbf{P}^T \widetilde{\mathbf{X}}_i) (\mathbf{P}^T \widetilde{\mathbf{X}}_i)^T \right]$$

Applying the matrix transposition rule $(\mathbf{A}\mathbf{B})^T = \mathbf{B}^T\mathbf{A}^T$:


$$\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \mathbb{E}\left[ \mathbf{P}^T \widetilde{\mathbf{X}}_i \widetilde{\mathbf{X}}_i^T \mathbf{P} \right]$$

Because the projection matrix $\mathbf{P}$ is deterministic, the linear expectation operator $\mathbb{E}[\cdot]$ moves inside to act directly on the random vector outer product:


$$\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \mathbf{P}^T \mathbb{E}[\widetilde{\mathbf{X}}_i \widetilde{\mathbf{X}}_i^T] \mathbf{P}$$

Substituting the baseline population covariance $\boldsymbol{\Sigma}$:


$$\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \mathbf{P}^T \boldsymbol{\Sigma} \mathbf{P}$$

Now, substituting the spectral decomposition $\boldsymbol{\Sigma} = \mathbf{P} \mathbf{\Lambda} \mathbf{P}^T$ into the expression:


$$\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \mathbf{P}^T (\mathbf{P} \mathbf{\Lambda} \mathbf{P}^T) \mathbf{P}$$

Using matrix associativity and the fundamental property of orthogonal matrices ($\mathbf{P}^T\mathbf{P} = \mathbf{I}$):


$$\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = (\mathbf{P}^T \mathbf{P}) \mathbf{\Lambda} (\mathbf{P}^T \mathbf{P}) = \mathbf{I} \cdot \mathbf{\Lambda} \cdot \mathbf{I} = \mathbf{\Lambda}$$

This simplifies exactly to the strictly diagonal eigenvalue matrix $\mathbf{\Lambda}$. $\blacksquare$

### Geometric and Statistical Interpretation

* **Statistical Meaning:** The off-diagonal elements of the transformed covariance matrix represent the cross-covariances between distinct principal coordinates: $\text{Cov}(Z_{i,j}, Z_{i,k}) = 0$ for all $j \neq k$. This means that the transformation completely **de-correlates** the features, stripping out all linear redundancies.
* **Geometric Meaning:** The original data coordinate axes have been rigidly rotated into a new orthogonal basis aligned with the directions of maximum variance. The features are now mutually perpendicular, meaning that moving along one principal axis provides exactly zero information about the location along any other axis.

## 2. Total Variance Conservation & Subspace Partitioning Ratio

### Formal Proof of Variance Conservation

The total variance of a multivariate system is defined as the sum of the variances of its individual channels, which corresponds algebraically to the trace ($\text{tr}$) of its covariance matrix.

We want to show that the total variance in the raw coordinate space equals the total variance in the principal component space:


$$\text{tr}(\boldsymbol{\Sigma}) = \text{tr}(\mathbf{\Lambda})$$

Starting with the spectral decomposition $\boldsymbol{\Sigma} = \mathbf{P} \mathbf{\Lambda} \mathbf{P}^T$, we apply the trace operator to both sides:


$$\text{tr}(\boldsymbol{\Sigma}) = \text{tr}(\mathbf{P} \mathbf{\Lambda} \mathbf{P}^T)$$

The cyclic permutation property of the trace states that for any matrices $\mathbf{A}$, $\mathbf{B}$, and $\mathbf{C}$ of appropriate dimensions:


$$\text{tr}(\mathbf{A}\mathbf{B}\mathbf{C}) = \text{tr}(\mathbf{C}\mathbf{A}\mathbf{B})$$

Letting $\mathbf{A} = \mathbf{P}$, $\mathbf{B} = \mathbf{\Lambda}$, and $\mathbf{C} = \mathbf{P}^T$, we cyclically permute $\mathbf{P}^T$ to the front of the product:


$$\text{tr}(\mathbf{P} \mathbf{\Lambda} \mathbf{P}^T) = \text{tr}(\mathbf{P}^T \mathbf{P} \mathbf{\Lambda})$$

Because $\mathbf{P}$ is an orthogonal matrix, $\mathbf{P}^T\mathbf{P} = \mathbf{I}_m$. Substituting this identity yields:


$$\text{tr}(\mathbf{P}^T \mathbf{P} \mathbf{\Lambda}) = \text{tr}(\mathbf{I}_m \mathbf{\Lambda}) = \text{tr}(\mathbf{\Lambda})$$

Because $\mathbf{\Lambda} = \text{diag}(\lambda_1, \lambda_2, \dots, \lambda_m)$, the trace simplifies to the sum of the population eigenvalues:


$$\text{tr}(\boldsymbol{\Sigma}) = \text{tr}(\mathbf{\Lambda}) = \sum_{j=1}^m \lambda_j$$

This proves that total system variance is preserved during the mapping. $\blacksquare$

### Subspace Partitioning Formulations

When truncating our model to retain $k$ components out of $m$ available features, the empirical parameters are structured as follows:

* **Cumulative Explained Variance Ratio $\Phi(k)$:** Measures the proportion of total systemic energy retained within the active principal subspace:

$$\Phi(k) = \frac{\sum_{j=1}^k \lambda_j}{\sum_{j=1}^m \lambda_j} = \frac{\text{tr}(\widehat{\mathbf{P}}_k^T \mathbf{S} \widehat{\mathbf{P}}_k)}{\text{tr}(\mathbf{S})}$$


* **Residual Unexplained Variance Ratio $\Psi(k)$:** Tracks the remaining noise floor fraction discarded into the residual null space:

$$\Psi(k) = \frac{\sum_{j=k+1}^m \lambda_j}{\sum_{j=1}^m \lambda_j} = 1 - \Phi(k)$$

## 3. Subspace Energy Decomposition & SHM Operational Diagnostics

### Vectors Definition

Given the partitioned realization $(\mathbf{x}_i - \widehat{\boldsymbol{\mu}}_n) = \widehat{\mathbf{P}}_k \mathbf{z}_{i,k} + \widehat{\mathbf{P}}_{m-k} \mathbf{z}_{i,m-k}$:

* **Reconstructed Vector ($\widehat{\mathbf{x}}_i$):** The projection of the sample back into the raw space using only the first $k$ dominant structural modes:

$$\widehat{\mathbf{x}}_i = \widehat{\boldsymbol{\mu}}_n + \widehat{\mathbf{P}}_k \mathbf{z}_{i,k}$$


* **Residual Error Vector ($\mathbf{e}_i$):** The unmodeled high-frequency noise or localized structural deviations pushed into the discarded subspace:

$$\mathbf{e}_i = (\mathbf{x}_i - \widehat{\boldsymbol{\mu}}_n) - (\widehat{\mathbf{x}}_i - \widehat{\boldsymbol{\mu}}_n) = \widehat{\mathbf{P}}_{m-k} \mathbf{z}_{i,m-k}$$



### Proof of Energy Decomposition via Pythagorean Identity

The total squared Euclidean norm of the centered observation is written as:


$$\|\mathbf{x}_i - \widehat{\boldsymbol{\mu}}_n\|^2 = (\mathbf{x}_i - \widehat{\boldsymbol{\mu}}_n)^T (\mathbf{x}_i - \widehat{\boldsymbol{\mu}}_n)$$

Substituting the partitioned subspace representation into the inner product:


$$= \left( \widehat{\mathbf{P}}_k \mathbf{z}_{i,k} + \widehat{\mathbf{P}}_{m-k} \mathbf{z}_{i,m-k} \right)^T \left( \widehat{\mathbf{P}}_k \mathbf{z}_{i,k} + \widehat{\mathbf{P}}_{m-k} \mathbf{z}_{i,m-k} \right)$$

Expanding the vector polynomial yields four terms:


$$= \mathbf{z}_{i,k}^T \widehat{\mathbf{P}}_k^T \widehat{\mathbf{P}}_k \mathbf{z}_{i,k} + \mathbf{z}_{i,k}^T \widehat{\mathbf{P}}_k^T \widehat{\mathbf{P}}_{m-k} \mathbf{z}_{i,m-k} + \mathbf{z}_{i,m-k}^T \widehat{\mathbf{P}}_{m-k}^T \widehat{\mathbf{P}}_k \mathbf{z}_{i,k} + \mathbf{z}_{i,m-k}^T \widehat{\mathbf{P}}_{m-k}^T \widehat{\mathbf{P}}_{m-k} \mathbf{z}_{i,m-k}$$

By definition of spectral decomposition, the partitioned principal matrices form mutually orthogonal and self-isometric subspaces:

1. **Subspace Self-Isometry:** $\widehat{\mathbf{P}}_k^T \widehat{\mathbf{P}}_k = \mathbf{I}_k$ and $\widehat{\mathbf{P}}_{m-k}^T \widehat{\mathbf{P}}_{m-k} = \mathbf{I}_{m-k}$
2. **Mutual Subspace Orthogonality:** $\widehat{\mathbf{P}}_k^T \widehat{\mathbf{P}}_{m-k} = \mathbf{0}_{k \times (m-k)}$ and $\widehat{\mathbf{P}}_{m-k}^T \widehat{\mathbf{P}}_k = \mathbf{0}_{(m-k) \times k}$

Applying these constraints to the expanded polynomial causes the two middle cross-subspace terms to vanish entirely:


$$= \mathbf{z}_{i,k}^T (\mathbf{I}_k) \mathbf{z}_{i,k} + \mathbf{0} + \mathbf{0} + \mathbf{z}_{i,m-k}^T (\mathbf{I}_{m-k}) \mathbf{z}_{i,m-k}$$

$$= \mathbf{z}_{i,k}^T \mathbf{z}_{i,k} + \mathbf{z}_{i,m-k}^T \mathbf{z}_{i,m-k}$$

Expressing the vector inner products back as squared Euclidean norms completes the proof:


$$\|\mathbf{x}_i - \widehat{\boldsymbol{\mu}}_n\|^2 = \|\mathbf{z}_{i,k}\|^2 + \|\mathbf{z}_{i,m-k}\|^2 \quad \blacksquare$$

## 1. The Covariance Decomposition Theorem & Variance Partitioning

### Proof of the Fundamental Equation of Factor Analysis

We are given the linear generative model structure:


$$\mathbf{Z}_i = \boldsymbol{\Lambda} \mathbf{F}_i + \boldsymbol{\epsilon}_i$$

The universal correlation matrix $\mathbf{R}$ is defined as the expectation of the outer product of the standardized vector $\mathbf{Z}_i$:


$$\mathbf{R} = \mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T]$$

Substituting the linear model equation into the outer product expression:


$$\mathbf{R} = \mathbb{E}\left[ (\boldsymbol{\Lambda} \mathbf{F}_i + \boldsymbol{\epsilon}_i) (\boldsymbol{\Lambda} \mathbf{F}_i + \boldsymbol{\epsilon}_i)^T \right]$$

Applying the matrix transposition identity $(\mathbf{A} + \mathbf{B})^T = \mathbf{A}^T + \mathbf{B}^T$ and expanding the polynomial yields:


$$\mathbf{R} = \mathbb{E}\left[ (\boldsymbol{\Lambda} \mathbf{F}_i + \boldsymbol{\epsilon}_i) (\mathbf{F}_i^T \boldsymbol{\Lambda}^T + \boldsymbol{\epsilon}_i^T) \right]$$

$$\mathbf{R} = \mathbb{E}\left[ \boldsymbol{\Lambda} \mathbf{F}_i \mathbf{F}_i^T \boldsymbol{\Lambda}^T + \boldsymbol{\Lambda} \mathbf{F}_i \boldsymbol{\epsilon}_i^T + \boldsymbol{\epsilon}_i \mathbf{F}_i^T \boldsymbol{\Lambda}^T + \boldsymbol{\epsilon}_i \boldsymbol{\epsilon}_i^T \right]$$

By the linearity of the expectation operator $\mathbb{E}[\cdot]$, we distribute the operator across each matrix term. Because the factor loading matrix $\boldsymbol{\Lambda}$ is a deterministic parameter, it factors out:


$$\mathbf{R} = \boldsymbol{\Lambda} \mathbb{E}[\mathbf{F}_i \mathbf{F}_i^T] \boldsymbol{\Lambda}^T + \boldsymbol{\Lambda} \mathbb{E}[\mathbf{F}_i \boldsymbol{\epsilon}_i^T] + \mathbb{E}[\boldsymbol{\epsilon}_i \mathbf{F}_i^T] \boldsymbol{\Lambda}^T + \mathbb{E}[\boldsymbol{\epsilon}_i \boldsymbol{\epsilon}_i^T]$$

Now we apply the structural constraints of the generative system:

1. **Orthogonal Common Factors:** $\mathbb{E}[\mathbf{F}_i \mathbf{F}_i^T] = \mathbf{I}_{k \times k}$
2. **Diagonalized Sensor Noise Floor:** $\mathbb{E}[\boldsymbol{\epsilon}_i \boldsymbol{\epsilon}_i^T] = \boldsymbol{\Psi}$
3. **Latent Orthogonality Assumptions:** The factors and unique errors are independent, implying $\mathbb{E}[\boldsymbol{\epsilon}_i \mathbf{F}_i^T] = \mathbf{0}_{m \times k}$. Taking the transpose confirms $\mathbb{E}[\mathbf{F}_i \boldsymbol{\epsilon}_i^T] = \mathbf{0}_{k \times m}$.

Substituting these identity matrices and zero matrices back into our expanded equation:


$$\mathbf{R} = \boldsymbol{\Lambda} (\mathbf{I}_{k \times k}) \boldsymbol{\Lambda}^T + \boldsymbol{\Lambda} (\mathbf{0}_{k \times m}) + (\mathbf{0}_{m \times k}) \boldsymbol{\Lambda}^T + \boldsymbol{\Psi}$$

$$\mathbf{R} = \boldsymbol{\Lambda} \boldsymbol{\Lambda}^T + \boldsymbol{\Psi} \quad \blacksquare$$

### Physical Interpretation of the $j$-th Diagonal Entry

Focusing on the $j$-th diagonal entry of the correlation matrix, the variance of a single standardized sensor (which equals $1.0$) partitions into two distinct scalar elements:


$$1.0 = R_{jj} = \left(\sum_{r=1}^k \lambda_{jr}^2\right) + \varphi_j^2$$

* **Communality ($h_j^2 = \sum_{r=1}^k \lambda_{jr}^2$):** The proportion of a sensor's variance that is explained by the shared latent factors.
* *SHM Physical Meaning:* This tracks the **global structural energy** felt by the sensor. High communality indicates that the instrument is successfully picking up systemic load behaviors, stress states, or operational modes shared with the rest of the asset network.


* **Uniqueness ($\varphi_j^2$):** The proportion of variance that is unique to that specific sensor channel, isolated from all others.
* *SHM Physical Meaning:* This defines the **local noise floor** or independent variance. A high uniqueness indicates that the sensor's readings are mostly driven by electrical noise, high-frequency localized vibration, or isolated hardware faults, rather than any global physical phenomenon.

## 2. Coordinate Transformations via Varimax Rotation

### Conceptual Contrasts: PCA vs. Varimax-Rotated FA

* **Raw PCA Loading Vectors:** Built using purely geometric optimization to maximize total global variance on each successive component. This mathematical constraint often forces cross-loadings, creating components where almost all sensors carry moderate weights. This mixes distinct physical phenomena together and makes troubleshooting difficult.
* **Varimax-Rotated FA Loading Vectors:** Relaxes the strict variance-hierarchy constraint to establish a **"simple structure."** The Varimax rotation drives loading values toward the extremes—either close to $1.0$ or close to $0.0$. Conceptually, this isolates the tracking channels so that each sensor aligns strongly with only one specific latent factor, providing a direct mapping between sensors and independent physical processes.

### Mathematical Invariance Under Orthogonal Rotations

Let $\mathbf{T}$ be a $k \times k$ orthogonal rotation matrix, which satisfies $\mathbf{T}\mathbf{T}^T = \mathbf{T}^T\mathbf{T} = \mathbf{I}_{k \times k}$. When we rotate the factor loading matrix, the new rotated coordinates become $\boldsymbol{\Lambda}_{\text{rotated}} = \boldsymbol{\Lambda}\mathbf{T}$.

* **Global Covariance Conservation:** Substituting the rotated loadings into the core decomposition equation shows that the estimated profile remains completely unchanged:

$$\boldsymbol{\Lambda}_{\text{rotated}} \boldsymbol{\Lambda}_{\text{rotated}}^T = (\boldsymbol{\Lambda}\mathbf{T})(\boldsymbol{\Lambda}\mathbf{T})^T = \boldsymbol{\Lambda}(\mathbf{T}\mathbf{T}^T)\boldsymbol{\Lambda}^T = \boldsymbol{\Lambda}(\mathbf{I})\boldsymbol{\Lambda}^T = \boldsymbol{\Lambda}\boldsymbol{\Lambda}^T$$



Because the cross-product eliminates $\mathbf{T}$, the global covariance approximation $\mathbf{R} \approx \boldsymbol{\Lambda}\boldsymbol{\Lambda}^T + \boldsymbol{\Psi}$ is perfectly preserved.
* **Communality Conservation:** The communality of the $j$-th sensor corresponds to the $j$-th diagonal element of the $\boldsymbol{\Lambda}\boldsymbol{\Lambda}^T$ matrix: $h_j^2 = [\boldsymbol{\Lambda}\boldsymbol{\Lambda}^T]_{jj}$. Because the matrix product $\boldsymbol{\Lambda}\boldsymbol{\Lambda}^T$ is invariant to the rotation matrix $\mathbf{T}$, its individual diagonal entries remain exactly the same.

> **Diagnostic Insight:** Orthogonal rotations redistribute the variance across the *columns* of the loading matrix (reallocating energy between factors to make them cleaner to interpret), but they never alter the total shared energy ($h_j^2$) or the distinct noise floor ($\varphi_j^2$) of any individual sensor.

## 3. Latent State Extraction via Thomson's Regression Method

### Partioned Joint Covariance Derivation

We construct the joint system vector $\mathbf{Y}_i = \begin{bmatrix} \mathbf{Z}_i \\ \mathbf{F}_i \end{bmatrix} \in \mathbb{R}^{m+k}$. Since the variables are zero-mean ($\mathbb{E}[\mathbf{Y}_i] = \mathbf{0}$), the block partitioned joint covariance matrix is evaluated as:


$$\boldsymbol{\Sigma}_{YY} = \mathbb{E}[\mathbf{Y}_i \mathbf{Y}_i^T] = \mathbb{E}\left[ \begin{bmatrix} \mathbf{Z}_i \\ \mathbf{F}_i \end{bmatrix} \begin{bmatrix} \mathbf{Z}_i^T & \mathbf{F}_i^T \end{bmatrix} \right] = \begin{bmatrix} \mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] & \mathbb{E}[\mathbf{Z}_i \mathbf{F}_i^T] \\ \mathbb{E}[\mathbf{F}_i \mathbf{Z}_i^T] & \mathbb{E}[\mathbf{F}_i \mathbf{F}_i^T] \end{bmatrix}$$

We evaluate each individual block submatrix inside this array:

1. **Top-Left Submatrix ($\boldsymbol{\Sigma}_{11}$):** By definition, this matches the universal correlation matrix: $\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \mathbf{R}$.
2. **Bottom-Right Submatrix ($\boldsymbol{\Sigma}_{22}$):** By definition of the orthogonal latent spaces: $\mathbb{E}[\mathbf{F}_i \mathbf{F}_i^T] = \mathbf{I}_{k \times k}$.
3. **Top-Right Cross-Covariance Submatrix ($\boldsymbol{\Sigma}_{12}$):** Substituting the generative equation for $\mathbf{Z}_i$:

$$\boldsymbol{\Sigma}_{12} = \mathbb{E}[\mathbf{Z}_i \mathbf{F}_i^T] = \mathbb{E}[(\boldsymbol{\Lambda}\mathbf{F}_i + \boldsymbol{\epsilon}_i)\mathbf{F}_i^T] = \boldsymbol{\Lambda}\mathbb{E}[\mathbf{F}_i\mathbf{F}_i^T] + \mathbb{E}[\boldsymbol{\epsilon}_i\mathbf{F}_i^T]$$



Applying the orthogonality assumptions ($\mathbf{I}$ and $\mathbf{0}$):

$$\boldsymbol{\Sigma}_{12} = \boldsymbol{\Lambda}(\mathbf{I}_{k \times k}) + \mathbf{0} = \boldsymbol{\Lambda}$$


4. **Bottom-Left Cross-Covariance Submatrix ($\boldsymbol{\Sigma}_{21}$):** This is simply the transpose of the top-right block:

$$\boldsymbol{\Sigma}_{21} = \mathbb{E}[\mathbf{F}_i \mathbf{Z}_i^T] = \boldsymbol{\Sigma}_{12}^T = \boldsymbol{\Lambda}^T$$



Assembling these four block calculations completes the matrix expansion:


$$\boldsymbol{\Sigma}_{YY} = \begin{bmatrix} \mathbf{R} & \boldsymbol{\Lambda} \\ \boldsymbol{\Lambda}^T & \mathbf{I}_{k \times k} \end{bmatrix} \quad \blacksquare$$

### Derivation of Thomson's Factor Estimator Formula

We want to extract the unobservable factor state realization using the conditional expectation $\mathbf{f}_i = \mathbb{E}[\mathbf{F}_i | \mathbf{z}_i]$.

The standard multivariate normal projection identity states that for a partitioned Gaussian vector, the conditional mean is given by:


$$\mathbb{E}[\mathbf{X}_2 | \mathbf{x}_1] = \boldsymbol{\mu}_2 + \boldsymbol{\Sigma}_{21} \boldsymbol{\Sigma}_{11}^{-1} (\mathbf{x}_1 - \boldsymbol{\mu}_1)$$

Mapping our joint system components directly onto this projection identity:

* Target random variable: $\mathbf{X}_2 \to \mathbf{F}_i$ (with mean $\boldsymbol{\mu}_2 = \mathbf{0}$)
* Conditioned observation: $\mathbf{x}_1 \to \mathbf{z}_i$ (with mean $\boldsymbol{\mu}_1 = \mathbf{0}$)
* Covariance entries: $\boldsymbol{\Sigma}_{11} = \mathbf{R}$ and $\boldsymbol{\Sigma}_{21} = \boldsymbol{\Lambda}^T$

Substituting these parameters directly into the conditional expectation projection formula:


$$\mathbf{f}_i = \mathbf{0} + (\boldsymbol{\Lambda}^T) (\mathbf{R})^{-1} (\mathbf{z}_i - \mathbf{0})$$

$$\mathbf{f}_i = \boldsymbol{\Lambda}^T \mathbf{R}^{-1} \mathbf{z}_i \quad \blacksquare$$

> **Operational Significance:** This final equation defines Thomson's Minimum Mean Squared Error (MMSE) estimator. It allows you to calculate the unobservable physical factor states $\mathbf{f}_i$ in real time by projecting incoming standardized sensor data $\mathbf{z}_i$ through the inverse correlation matrix and factor loading weights.

## The Mathematical Mechanics of the Illusion

When you feed raw or standard-scaled sensor streams into PCA, the engine maximizes the objective function:

$$\max_{\mathbf{p}} \mathbf{p}^T \mathbf{S} \mathbf{p} \quad \text{subject to} \quad \mathbf{p}^T\mathbf{p} = 1$$

If a single broken instrument (`Sensor_4`) undergoes an electrical short or a thermal runaway, its independent local variance spikes dramatically. Because PCA treats all variance identically—whether it is driven by global structural modes or isolated high-frequency hardware noise—the optimization layer is forced to rotate the primary coordinate axis (`PC 1`) directly toward that outlier sensor to capture its energy.
